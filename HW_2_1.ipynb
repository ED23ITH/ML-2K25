Task 2.12:Write a program that prints a simple number pyramid pattern using nested loops.

```python
n=int(input("Enter the number of rows:"))
for i in range (n):
  for j in range (n-i-1):
    print(" ",end="")
  for j in range (2*i-1):
    print("*",end="")
  print()
```
